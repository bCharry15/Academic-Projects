# Inputación de Datos

Inputación de Datos

Imputación de Datos
Los métodos de imputación consisten en estimar los valores ausentes en base a los valores válidos de otras variables y/o casos de la muestra. La estimación se puede hacer a partir de la información del conjunto completo de variables o bien de algunas variables especialmente seleccionadas

#### 9.1 Imputación de Variables Numéricas (con la Mediana)

```python
# Imputar valores faltantes en df_numericas con la mediana
for col in df_numericas.columns:
    if df_numericas[col].isnull().any():
        median_val = df_numericas[col].median()
        df_numericas[col] = df_numericas[col].fillna(median_val)

print("Valores nulos en df_numericas después de la imputación:")
display(df_numericas.isnull().sum())
```

#### 9.2 Imputación de Variables Categóricas (con la Moda)

```python
# Imputar valores faltantes en df_categoricas con la moda
for col in df_categoricas.columns:
    if df_categoricas[col].isnull().any():
        mode_val = df_categoricas[col].mode()[0]
        df_categoricas[col] = df_categoricas[col].fillna(mode_val)

print("Valores nulos en df_categoricas después de la imputación:")
display(df_categoricas.isnull().sum())
```

# Codificar One Hot Encoder


Cargo los valores del LET_IS

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Lista de columnas a codificar
columnas_codificar = ['LET_IS']

# Crear el codificador
encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown='ignore'
)

# Aplicar One Hot Encoding
datos_codificados = encoder.fit_transform(
    df_categoricas[columnas_codificar]
)

# Convertir el resultado en DataFrame
df_codificado = pd.DataFrame(
    datos_codificados,
    columns=encoder.get_feature_names_out(columnas_codificar),
    index=df_categoricas.index
)

# Eliminar LET_IS original y guardar el resultado en un nuevo DataFrame
df_categoricas_codificado = df_categoricas.drop(columns=['LET_IS'])

# Agregar las nuevas columnas codificadas
df_categoricas_codificado = pd.concat(
    [df_categoricas_codificado, df_codificado],
    axis=1
)

display(df_categoricas_codificado.head())
```

```python
X_imp = pd.concat([
    df_numericas,
    df_categoricas.drop(columns=['LET_IS'])
], axis=1)

y_imp = df_categoricas['LET_IS']

print(X_imp.shape)
print(y_imp.shape)
```

```python
from sklearn.model_selection import train_test_split

X_imp_train, X_imp_test, y_imp_train, y_imp_test = train_test_split(
    X_imp,
    y_imp,
    test_size=0.33,
    random_state=0,
    stratify=df_categoricas['LET_IS'],
    shuffle=True
)

print("X_imp_train:", X_imp_train.shape)
print("X_imp_test:", X_imp_test.shape)
print("y_imp_train:", y_imp_train.shape)
print("y_imp_test:", y_imp_test.shape)
```
